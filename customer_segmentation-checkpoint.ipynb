```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("smartcart_customers.csv")
```

```python
from sklearn.impute import SimpleImputer
si = SimpleImputer(strategy = 'median')
data['Income']=si.fit_transform(data[['Income']])
```

# Feature Engineering

```python
data['Dt_Customer'] = pd.to_datetime(data['Dt_Customer'],dayfirst=True)
reference_data = data['Dt_Customer'].max()
data['Customer_Tenure_Days'] = (reference_data - data['Dt_Customer']).dt.days
```

```python
data['Total_AMT_Spent'] = data['MntWines']+data['MntFruits']+data['MntMeatProducts']+data['MntFishProducts']+data['MntSweetProducts']+data['MntGoldProds']
```

```python
education_map = {
    'Graduation': 'Graduate',
    'Basic': 'Graduate',
    '2n Cycle': 'Graduate',
    'Master': 'Post_Graduate',
    'PhD': 'Post_Graduate'
}
data['Education'] = data['Education'].map(education_map)
```

```python
data['Education'].unique()
```

```python
data['Total_Children'] = data['Kidhome'] + data['Teenhome']
```

```python
marital_status = {
    'Single': 'Alone',
    'Divorced': 'Alone',
    'Widow': 'Alone',
    'Alone': 'Alone',
    'Absurd': 'Alone',
    'YOLO': 'Alone',
    'Together': 'Together',
    'Married': 'Together'
}
data['Living'] = data['Marital_Status'].map(marital_status)
```

```python
data['Age'] = 2026 - data['Year_Birth']
```

```python
data.head()
```

# Outlier

```python
# cols = ['Income','Recency','Total_Children','Age','Total_AMT_Spent'] 
# sns.pairplot(data[cols])
```

```python
data = data[data['Income'] < 600000]
data = data[data['Age'] <= 90]
```

```python
# sns.pairplot(data[cols])
# plt.show()
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()
encoded_data = ohe.fit_transform(data[['Education','Living']]).toarray()
```

```python
ohe.get_feature_names_out()
```

```python
columns = ohe.get_feature_names_out()
cleaned_data = pd.concat([
    data.drop(columns=[
        'ID',
        'Education',
        'Marital_Status',
        'Kidhome',
        'Teenhome',
        'Dt_Customer',
        'MntWines',
        'MntFruits',
        'MntMeatProducts',
        'MntFishProducts',
        'MntSweetProducts',
        'MntGoldProds',
        'Marital_Status',
        'Living',
        'Year_Birth'
    ]),
    pd.DataFrame(encoded_data,columns=columns,index=data.index)
],axis=1)
```

```python
cleaned_data.head()
```

## Heatmap

```python
corr = cleaned_data.corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr,annot=True,annot_kws={'size':6},cmap='coolwarm')
```

## Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(cleaned_data)
```

## visualize

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=3)
pca_data = pca.fit_transform(scaled_data)
```

```python
pca.explained_variance_ratio_
```

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111,projection='3d')
ax.scatter(pca_data[:,0],pca_data[:,1],pca_data[:,2])
ax.set_xlabel('PCA1')
ax.set_ylabel('PCA2')
ax.set_zlabel('PCA3')
ax.set_title('3D Projection')
plt.show()
```

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1,11):
    kmean=KMeans(n_clusters=i)
    kmean.fit_predict(pca_data)
    wcss.append(kmean.inertia_)
```

```python
from kneed import KneeLocator
knee = KneeLocator(range(1,11),wcss,curve='convex',direction='decreasing')
knee.elbow
```

```python
sns.lineplot(x=range(1,11),y=wcss,marker='o')
plt.xlabel('Range 1 - 11')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.grid()
plt.show()
```

### Silhouette Score 

```python
from sklearn.metrics import silhouette_score
scores = []
for i in range(2,11):
    kmean = KMeans(n_clusters=i)
    labels = kmean.fit_predict(pca_data)
    score = silhouette_score(pca_data,labels)
    scores.append(score)
```

```python
sns.lineplot(x=range(2,11),y=scores,marker='o')
plt.xlabel('Range 2 - 11')
plt.ylabel('scores')
plt.title('silhouette score')
plt.grid()
plt.show()
```

```python
# combine Plot

k_range = range(2,11)
fig,ax1 = plt.subplots(figsize=(10,8))
ax1.plot(k_range,wcss[:len(k_range)],marker='o',c='blue')
ax1.set_xlabel('K')
ax1.set_ylabel('wcss')
ax2 = ax1.twinx()
ax2.plot(k_range,scores[:len(k_range)],marker='x',linestyle='--',c='red')
ax2.set_ylabel('score')
plt.grid()
```

```python
kmean = KMeans(n_clusters=4)
km_label = kmean.fit_predict(pca_data)
```

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111,projection='3d')
ax.scatter(pca_data[:,0],pca_data[:,1],pca_data[:,2],c=km_label)
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("KMeans clusters")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
agg = AgglomerativeClustering(n_clusters=4)
agg_labels = agg.fit_predict(pca_data)
```

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111,projection='3d')
ax.scatter(pca_data[:,0],pca_data[:,1],pca_data[:,2],c=agg_labels)
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("KMeans clusters")
plt.show()
```

```python
cleaned_data['clusters'] = km_label
```
